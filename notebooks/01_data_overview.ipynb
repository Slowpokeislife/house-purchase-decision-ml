# House Purchase Decision — Clean EDA

This notebook is **new**, based on the study design but without hidden outputs, usernames, OAuth URLs or copied raw training data. Requires `house-purchase download`. All stats are computed when executed locally. Avoid sharing individual rows.

```python
from house_purchase.data import load_data, split_data
from house_purchase.config import DEFAULT_DATA, FEATURES, CATEGORICAL, NUMERICAL

df = load_data(DEFAULT_DATA)
print('Rows, columns:', df.shape)
print('Class distribution (%)')
print((df['decision'].value_counts(normalize=True) * 100).round(2))
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(10, 3))
df['decision'].value_counts().sort_index().plot(kind='bar', ax=axes[0], title='Target balance')
df['property_size_sqft'].plot(kind='hist', bins=40, ax=axes[1], title='Property size')
plt.tight_layout()
```

## Correct split
The original notebook fitted the encoders/scaler on the full dataset. Here, we split first and only fit preprocessing on training rows. Validation is for training/Optuna; test is a final holdout.

```python
from house_purchase.data import make_preprocessor

s = split_data(df, seed=42)
preprocessor = make_preprocessor()
X_train = preprocessor.fit_transform(s.X_train)
X_val = preprocessor.transform(s.X_val)
X_test = preprocessor.transform(s.X_test)
print('Train/validation/test:', X_train.shape, X_val.shape, X_test.shape)
```
